```python
import pandas as pd
```

```python
data = {
    'Employee_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109],
    'Department': ['IT', 'IT', 'IT', 'IT', 'IT',
                   'HR', 'HR', 'HR', 'HR'],
    'Age': [25, 28, 30, 32, 35, 26, 29, 31, 40],
    'Salary': [30000, 32000, 31000, 35000, 500000,
               25000, 27000, 28000, 400000]
}
```

```python
df=pd.DataFrame(data)
```

```python
df.head(6)
```

```python
for Department in df['Department'].unique():
  dept_data=df[df['Department']== Department]

  Q1=dept_data['Salary'].quantile(0.25)
  Q3=dept_data['Salary'].quantile(0.75)
  IQR=Q3-Q1
print(Q1)
print(Q3)
print(IQR)

```

```python
lower_limit= Q1-1.5 * IQR
upper_limit= Q3+1.5 * IQR
```

```python
print(lower_limit)
print(upper_limit)
```

```python
outliers=dept_data[
    (dept_data['Salary']<lower_limit) | (dept_data['Salary']>upper_limit)
]
```

```python
print(outliers)
```

```python
if outliers.empty:
    print("No outliers found")
else:
    print(outliers[['Employee_ID', 'Department', 'Age',
                    'Salary']].to_string(index=False))
```
